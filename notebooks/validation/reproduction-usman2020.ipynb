# Reproduced experiment: ⁴He + ¹⁶⁵Ho, Usman et al. 2020

A. R. Usman et al., Nucl. Instrum. Methods B 469, 42 (2020) (EXFOR E2650): ¹⁶⁵Ho(α,xn)Tm by stacked-foil activation at the RIKEN AVF cyclotron, 26 Ho foils from 49.6 to 13.2 MeV, reproduced end to end through the planner. For each foil the `fusion_tm165` example is set to the foil's energy, its channel table filled from physim's own fusion and evaporation ("Fill from physim"), the residues simulated, and the production cross sections inferred from the residues made (`physim.nuclear.reproduction`). Kormazeva et al. 2022 and Gadkari et al. 1997 are the cross-checks; PACE4 4.36.0 ran at the same energies. The validation check `reproduction_usman2020` reruns the reproduction and judges it; this page reads the saved result, `tests/reference/nuclear/reproduction_usman2020.csv` (`scripts/reproduce_usman2020.py`).

```python
from physim.nuclear import reproduction as rp

fig = rp.figure()
```

physim's band is its statistical error. Judged above 1.1 V_B: the peaks of 2n, 3n and 4n (energy spread + 1 MeV), the strongest channel at each foil where the measurement resolves it, and the strongest three channels within 30% of the band of the three measurements up to each channel's peak. Above the peaks the measured tails are fed by pre-equilibrium emission, which neither physim nor PACE4 models: reported, not judged.

```python
from IPython.display import Markdown

c = rp.compare(rp.saved(), rp.measured())
lines = ['| Comparison | Tolerance | physim | measured | Result |', '|---|---|---|---|---|']
for x in c['peaks']:
    res = ('pass' if x['passed'] else 'fail') if x['judged'] else 'reported'
    lines.append(f"| {x['channel']} peak | ±{x['tolerance_mev']:.1f} MeV | {x['physim_mev']:.1f} MeV | "
                 f"{x['measured_mev']:.1f} MeV | {res} |")
s = [x for x in c['strongest'] if x['judged']]
lines.append(f"| strongest channel | the same | {sum(x['passed'] for x in s)} of {len(s)} foils | | "
             f"{'pass' if c['ok']['strongest'] else 'fail'} |")
m = [x for x in c['magnitudes'] if x['judged']]
w = max(m, key=lambda x: abs(x['deviation']))
lines.append(f"| magnitudes ({len(m)} points) | 30% of the band | worst {w['deviation']:+.0%} "
             f"({w['channel']}, {w['e_lab_mev']:g} MeV) | | {'pass' if c['ok']['magnitudes'] else 'fail'} |")
for x in c['ratio_3n_4n']:
    lines.append(f"| σ(3n)/σ(4n), {x['ref']} {x['e_lab_mev']:g} MeV | reported | {x['physim']:.3f} | "
                 f"{x['measured']:.3f} | pre-equilibrium |")
Markdown('\n'.join(lines) + '\n\n' + rp.reported(c) + '.')
```
